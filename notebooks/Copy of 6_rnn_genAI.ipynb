```python
# Dataset class comes from torch.utils.data
class CharDataset(Dataset):
    def __init__(self, data, block_size=8):
        self.data = data
        self.block_size = block_size

    def __len__(self):
        return len(self.data) - self.block_size

    def __getitem__(self, idx):
      # x: returns slice of 8 start at idx. y is also a slice but staring at idx+1
        x = self.data[idx:idx+self.block_size]
        y = self.data[idx+1:idx+self.block_size+1]
        return x, y
```

```python
block_size = 8
dataset = CharDataset(encoded_text, block_size)
print("getting block starting at idx 0:\n", dataset.__getitem__(0))

# the loader object "wraps" the dataset
loader = DataLoader(dataset, batch_size=16, shuffle=True)
```

```python
#######################################################################
# Our main class: RNN model , builds on top of the pre-existing class nn.Module
# this defines the architecture of the RNN neural net

class CharRNN(nn.Module):
    def __init__(self, vocab_size, emb_size=32, hidden_size=64):
        super().__init__()

        # the 1st layer is an embedding layer
        self.embed = nn.Embedding(vocab_size, emb_size)

        # this is the MEAT of the neural net.
        self.rnn = nn.RNN(emb_size, hidden_size, batch_first=True)

        # the final layer from hidden_size (64) neurons to vocab_size neurons
        self.fc = nn.Linear(hidden_size, vocab_size)

        self.printed = False

        self.numForwards = 0


    # the forward function receives a tensor (x) [think vector]
    # and passes it thru the recurrent neural net's layers
    def forward(self, x):

        if not self.printed:
          print("x shape: ", x.shape)

        x = self.embed(x)

        if not self.printed:
          print("embed shape: ", x.shape)

        out, _ = self.rnn(x)
        if not self.printed:
          print("output of rnn: ", out.shape)

        logits = self.fc(out)
        if not self.printed:
          print("logits.shape: ", logits.shape)

        self.printed = True
        self.numForwards += 1

        # for each char in vocab, there is a logit number
        # greater logit indicates the net thinks that that character is more
        # likely to be the next char
        return logits

    def getNumForwards(self):
      return self.numForwards
```

```python
#######################################################################
# Defining the loss function and optimizer
device = "cuda" if torch.cuda.is_available() else "cpu"
print("device:", device)

# tell the model object to connect itself to the appropriate device
model = CharRNN(vocab_size).to(device)

# lets decide what kind of loss function to use
criterion = nn.CrossEntropyLoss()

# lets decide what kind of optimizer (Adam) we will use and the
# learning rate (lr) is set to 0.01
optimizer = optim.Adam(model.parameters(), lr=0.01)
```

```python
#######################################################################
# Training loop (5 epochs)

def learn():
  for epoch in range(10):
      for xb, yb in loader:
          xb, yb = xb.to(device), yb.to(device)
          logits = model(xb)
          loss = criterion(logits.view(-1, vocab_size), yb.view(-1))

          optimizer.zero_grad()
          loss.backward()
          optimizer.step()
          if loss.item() < 0.2 :
            return loss, model, epoch

      if epoch % 1 == 0:
        print(f"Epoch {epoch+1}: loss = {loss.item():.4f}, numForwards={model.getNumForwards()}")
  return loss, model, epoch

loss, model, epoch = learn()
print(f"Final Epoch {epoch+1}: loss = {loss.item():.4f}, numForwards={model.getNumForwards()}")
```

```python
# generate accepts a starting char (e.g. T) and ideally generates
# a sentence in the style of Shakespeare, we will ask for 20
# next characters
def generate(model, start_char='T', maxCharsNeeded=20):

    # set the model into eval mode (as opposed to training mode)
    model.eval()

    printed = False

    # take the starting char, convert it to its index (a long integer)
    # create a tensor out of that integer and send the tensor to the
    # device (cpu or gpu)
    idx = torch.tensor([[stoi[start_char]]], dtype=torch.long).to(device)

    # initialize the output array
    output = [start_char]
    hidden = None

    # get the next N chars as predicted by RNN
    for _ in range(maxCharsNeeded):

        # run the model on the current char (as an integer)
        logits = model(idx)
        if not printed:
          print("shape of logits: ", logits.shape)
          print("output of model: ", logits)

        # convert logits to probabilities
        probs = torch.softmax(logits[:, -1, :], dim=-1)
        if not printed:
          print("shape after slicing: ", logits[:, -1, :].shape)
          print("probs shape: ", probs.shape)
          for i in range(len(chars)):
            print(i, repr(itos[i]), probs[0, i])

        # SAMPLE from the probs distribution to get the
        # index of the next predicted char
        next_idx = torch.multinomial(probs, num_samples=1)

        # map from an index to the predicted char itself
        next_char = itos[next_idx.item()]

        # append the predicted char to the output array
        output.append(next_char)

        # store next_idx into idx so loop can start over
        idx = next_idx

        printed = True

    # convert the output array into a string
    return ''.join(output)
```
